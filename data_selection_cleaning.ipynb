# Selecting and cleaning data

```python
import pandas as pd

raw_csv = pd.read_csv("311_Service_Requests.csv")
print(raw_csv.head(3))
print(raw_csv.shape)
```

### Dealing with the warning above:

```python
# print(raw_csv.columns, "------", raw_csv.columns[35])

### drop all columns which are >95% missing values
high_na_values = []
for c in raw_csv.columns: 
    if raw_csv[c].notna().sum() < 0.05 * len(raw_csv.index): 
        high_na_values.append(c)

print(high_na_values)

trimmed_311 = raw_csv.drop(columns=high_na_values)
print(trimmed_311.head(3))
```

We can see that most of the columns with >95% missing values are irrelevant in the context of explaining long case-closure time. I feel confident in omitting these columns. One column with >50% missing data is the `Additional Details` column, but since this column may serve to be useful, I will keep it.
